1. **What is the difference between a function and a method in Python?**
  
  - In Python, both functions and methods are blocks of code that perform a specific task, but they differ mainly in how they are used and where they are defined.
  
    - **Function**: A function is defined using the def keyword and is not tied to any object. It can be called independently.
    
      - **Exanple**

```python
def Greet (name):
  return "Hello", name

Greet("rahul")
```

    -  **Method**: A method is also defined using the def keyword but is associated with an object. It is called on an instance of a class and can access the data within that object.

      - **Example**

```python
class Greeting:
  def handshake(self, name):
    return"Greeting", name

G1 = Greeting()
G1.handshake("rahul")
```

2. **Explain the concept of function arguments and parameters in Python.**
     -  In Python, parameters and arguments are used to pass data into functions. Though often used interchangeably, they have distinct meanings:
       
       - **Parameters** are the variable names listed inside the parentheses in the function definition.

       - **Arguments** are the actual values passed to the function when it is called.

- Example -

```python
def add(a, b):  # a and b are parameters
    return a + b

result = add(3, 5)  # 3 and 5 are arguments
print(result)  # Output: 8
```

3. **What are the different ways to define and call a function in Python?**
     -  In Python, a function is a reusable block of code that performs a specific task. Functions can be defined and called in various ways depending on the use case.

       -  1. **Standard Function Definition** - Defined using the def keyword with optional parameters.
       -  2. **Function with Default Arguments** - Provides default values for parameters.
       -  3. **Function with Variable-Length Arguments** - Allows functions to accept any number of arguments.
       -  4. **Lambda (Anonymous) Function** - Used for simple operations. Defined using the lambda keyword.
       -  5. **Calling a Function Returned by Another Function** - Functions can return other functions.

```python
#Standard Function Definition Example.
def Greet (name):
  return "Hello", name

Greet("rahul")
```

```python
#Function with Default Arguments Example.
def Greet (name = "ram"):
  print("Hello", name)

print(Greet, "rahul")
```

```python
#Function with Variable-Length Arguments Example.
def total(*numbers):
    print(sum(numbers))

total(1, 2, 3)  # Positional variable-length

def show_info(**details):
    print(details)

show_info(name="Ram", age=25)  # Keyword variable-length
```

```python
#Lambda (Anonymous) Function Example.
square = lambda x: x * x
print(square(4))  # Output: 16
```

```python
#Calling a Function Returned by Another Function Example.
def outer():
    def inner():
        print("Inside inner function")
    return inner

func = outer()
func()  # Calls inner function
```

6. **Explain the concept of generators in Python and how they are defined.**
    -  In Python, a generator is a special type of iterator that allows you to iterate over data lazily, meaning it generates items one at a time only when needed. This makes generators memory-efficient, especially for large datasets.
       -  **How Generators Are Defined** - Generators are defined like normal functions, but instead of using return, they use the yield keyword to produce a series of values.

**Example** -

```python
def count_up_to(n):
    count = 1
    while count <= n:
        yield count
        count += 1

for number in count_up_to(3):
    print(number)
```

7. **What are the advantages of using generators over regular functions?**
     - Generators in Python offer several advantages, especially when working with large datasets or streams of data. Unlike regular functions that return all results at once using return, generators use yield to produce one value at a time, which makes them more efficient in certain cases.
      - **Advantages** -
         1. Generators do not store all values in memory. They generate items one by one, which is ideal for processing large files or datasets.
         2. Generators don’t compute all values up front, so they start running immediately.
         3. Values are only generated when requested, reducing unnecessary computation.
         4. Generators can simplify code that involves iteration, especially when using loops or managing states.
         5. Generators are well-suited for representing infinite sequences, like a stream of numbers or sensor data.

     - **Purpose:-**  Automates the process of applying a function to every item in a sequence. Helps avoid explicit loops, making code shorter and more readable.

10. **What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?**
     - In Python, map(), reduce(), and filter() are built-in functions used for functional-style programming. Each serves a different purpose in processing data from iterables like lists or tuples.

      - **map()** - ***Purpose*** Applies a function to each item in an iterable and returns a new iterable (map object).and it's Use Case is Transforming data.

      - **filter()** -  **Purpose**: Applies a function to each item and keeps only items where the function returns True. and it's Use Case is Filtering data.

      -  **reduce()**  - **Purpose**: Applies a function cumulatively to the items, reducing the iterable to a single value. and it's Use Case is Aggregating data.

1. **Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in the list.**

```python
def sum_of_evens(numbers):
    return sum(num for num in numbers if num % 2 == 0)

my_list = [1, 2, 3, 4, 5, 6]
result = sum_of_evens(my_list)
print(result)
```

4. **Write a Python function that checks if a given number is prime or not from 1 to 200.**

```python
def is_prime(n):
    if n < 2 or n > 200:
        return False  # Only check primes between 1 and 200
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

print(is_prime(2))
print(is_prime(100))
print(is_prime(199))
print(is_prime(201))
```

5. **Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.**

```python
class FibonacciIterator:
    def __init__(self, max_terms):
        self.max_terms = max_terms
        self.count = 0
        self.a, self.b = 0, 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.max_terms:
            raise StopIteration
        if self.count == 0:
            self.count += 1
            return 0
        elif self.count == 1:
            self.count += 1
            return 1
        self.a, self.b = self.b, self.a + self.b
        self.count += 1
        return self.a

fib = FibonacciIterator(5)
for num in fib:
    print(num)
```

6. **Write a generator function in Python that yields the powers of 2 up to a given exponent.**


```python
def powers_of_2(exponent):
    for i in range(exponent + 1):
        yield 2 ** i
for power in powers_of_2(5):
    print(power)
```

10. **Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.**

```python
# List of temperatures in Celsius
celsius_temps = [0, 20, 25, 30, 40, 100]

# Function to convert Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

# Using map() to apply the conversion function to each element in the list
fahrenheit_temps = list(map(celsius_to_fahrenheit, celsius_temps))

# Printing the list of Fahrenheit temperatures
print(fahrenheit_temps)
```
